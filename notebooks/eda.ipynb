Load dataset

```python
df = pd.read_csv('../data/Boston House Price Data.csv')
df
```

```python
df.describe()
```

Correlation between features

```python
correlation = df.corr()
```

```python
# correlation heat map
plt.figure(figsize=(10, 10))
new = sns.heatmap(correlation, cbar=True, square=True, fmt='.1f', annot=True, annot_kws={'size' : 8}, cmap='Blues')
new.set_yticklabels(new.get_yticklabels(), rotation=0)
plt.show()
```

Feature correlation with price

```python
df.corr()['PRICE'].drop('PRICE').sort_values().plot(kind='barh', figsize=(8, 6))
plt.show()
```

```python
features = ['RM', 'LSTAT', 'PTRATIO', 'INDUS', 'TAX', 'NOX']
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, col in zip(axes.flatten(), features):
    sns.regplot(x=df[col], y=df['PRICE'], ax=ax, scatter_kws={'alpha': 0.4}, line_kws={'color': 'red'})
plt.tight_layout()
plt.show()
```

```python
df.hist(bins=30, figsize=(16, 12), edgecolor='black')
plt.tight_layout()
plt.show()
```
